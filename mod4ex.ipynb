```python
import pandas as pd
```

```python
## read data
data = pd.read_csv('QADSMLP-learner/data/renfe_trains.csv')
```

```python
## data in dataframe to visualise properly
df = pd.DataFrame(data)
```

```python
print(df)
```

```python
## inspect columns
df.info()
```

```python
## there are some values in price column with value "price". .value_counts() is used to visualise them
df['price'].value_counts()
```

```python
## inspecting the rows that have the value price via masking
## df['price'] == 'price' creates a boolean series for each row (true or false)
## df[boolean_series] keeps rows where boolean is true and drops rows where the boolean is false
df[df['price'] == 'price']
```

```python
## drop rows
df = df[df['price'] != 'price']
```

```python
import numpy as np
```

```python
## converting price to appropriate type
## astype() is a pandas method (not numpy), but it uses numpy data types like np.float32
df['price'] = df['price'].astype(np.float32)
```

```python
# mv = df.isnull
mv = df.isna().any
```

```python
mv
```

```python
## finding null values in vehicle class, price and fare
df.isna().sum()
```

```python
df[df['price'].isna()]
```

```python
## drop all rows which have missing vehicle_class, price and fare
df.dropna(subset = ['vehicle_class', 'price', 'fare'], how='all', inplace=True)
```

```python
df[['vehicle_class', 'fare', 'price']].groupby(['vehicle_class', 'fare']).mean()
```

```python
## fill the remaining price values with mean of all prices
df.fillna({'price': df['price'].mean()}, inplace=True)
```

```python
## Deduplication
df.duplicated()
```

```python
## drop duplicates. Use inplace=True to drop them inplace
df.drop_duplicates(inplace=True)
```

```python
df.duplicated()
```

```python
## outliers
q1= np.percentile(df['price'], 25)
```

```python
q3 = np.percentile(df['price'], 75)
```

```python
iqr = q3 - q1
```

```python
lower_bound = q1 - 1.5 * iqr
upper_bound = q3 - 1.5 * iqr
```

```python
outliers = [x for x in df['price'] if x < lower_bound or x > upper_bound]
```

```python
print(outliers)
```
