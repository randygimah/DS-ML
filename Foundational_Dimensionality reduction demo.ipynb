```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
from sklearn import datasets
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE
import seaborn as sns
import pandas as pd

# Set a style for plots
plt.style.use('seaborn-v0_8-darkgrid')
%matplotlib inline
```

```python
n_samples = 1500
X, color = datasets.make_swiss_roll(n_samples, noise=1.0, random_state=42)

fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')
ax.scatter(X[:, 0], X[:, 1], X[:, 2], c=color, cmap=plt.cm.Spectral, s=30)
ax.set_title("Original 3D Swiss Roll Data")
ax.set_xlabel("Feature 1")
ax.set_ylabel("Feature 2")
ax.set_zlabel("Feature 3")
plt.show()
```

```python
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X)

plt.figure(figsize=(10, 8))
plt.scatter(X_pca[:, 0], X_pca[:, 1], c=color, cmap=plt.cm.Spectral, s=30)
plt.title("Swiss Roll Data after PCA (2D)")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.colorbar(label="Original Color/Value")
plt.show()
```

```python
explained_variance_ratio = pca.explained_variance_ratio_
print(f"Explained variance ratio by PC1: {explained_variance_ratio[0]:.2f}")  #Explained variance ratio answers: What proportion of the total variance in the data is explained by each principal component?
print(f"Explained variance ratio by PC2: {explained_variance_ratio[1]:.2f}")
print(f"Total explained variance by 2 components: {sum(explained_variance_ratio):.2f}")

plt.figure(figsize=(8, 6))
plt.bar(range(1, len(explained_variance_ratio) + 1), explained_variance_ratio, alpha=0.7)
plt.ylabel('Proportion of Variance Explained')
plt.xlabel('Principal Component')
plt.title('Explained Variance by Principal Components')
plt.xticks(range(1, len(explained_variance_ratio) + 1))
plt.show()
```
